```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import os

# Unzip the dataset first
!unzip /content/skin-cancer-mnist-ham10000.zip

# Load the CSV file (28x28 RGB version)
df = pd.read_csv('hmnist_28_28_RGB.csv')

# Basic information
print("Dataset Shape:", df.shape)
print("\nColumn Info:")
print(df.info())
print("\nFirst few rows:")
df.head()
```

```python
X = df.drop('label', axis=1)
y = df['label']

print(f"Total samples: {len(df)}")
print(f"Number of features (pixels): {X.shape[1]}")
print(f"Image dimensions: 28 x 28 x 3 (RGB) = {28*28*3} pixels")
print(f"Number of classes: {y.nunique()}")

# Class distribution
class_names = {0: 'akiec', 1: 'bcc', 2: 'bkl', 3: 'df', 4: 'mel', 5: 'nv', 6: 'vasc'}
print("\nClass Distribution:")
print(y.value_counts().sort_index().rename(index=class_names))
```

```python
def display_samples(df, class_names, samples_per_class=3):
    fig, axes = plt.subplots(7, samples_per_class, figsize=(12, 20))

    for idx, (class_id, class_name) in enumerate(class_names.items()):
        class_samples = df[df['label'] == class_id].sample(samples_per_class)

        for j, (_, row) in enumerate(class_samples.iterrows()):
            pixels = row.drop('label').values.reshape(28, 28, 3).astype(np.uint8)
            axes[idx, j].imshow(pixels)
            axes[idx, j].set_title(f'{class_name}')
            axes[idx, j].axis('off')

    plt.tight_layout()
    plt.savefig('sample_images.png', dpi=150)
    plt.show()

display_samples(df, class_names)
```

```python
# Missing values
print("Missing values per column:")
print(df.isnull().sum().sum())

# Duplicates
duplicates = df.duplicated()
print(f"\nDuplicate rows: {duplicates.sum()}")

# Remove duplicates if any
df_clean = df.drop_duplicates()
print(f"Shape after removing duplicates: {df_clean.shape}")
```

```python
# Check data types
print("Data types:")
print(df.dtypes.value_counts())

# Verify pixel value range
X = df_clean.drop('label', axis=1)
print(f"\nPixel value range: [{X.min().min()}, {X.max().max()}]")
```

```python
# Normalize to [0, 1] range
X_normalized = X.values / 255.0

# Reshape to image format (samples, height, width, channels)
X_reshaped = X_normalized.reshape(-1, 28, 28, 3)

print(f"Reshaped data: {X_reshaped.shape}")
print(f"Normalized range: [{X_reshaped.min()}, {X_reshaped.max()}]")
```

```python
from sklearn.model_selection import train_test_split

y = df_clean['label'].values

# Stratified split to maintain class distribution
X_train, X_test, y_train, y_test = train_test_split(
    X_reshaped, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")

# Verify stratification
print("\nClass distribution in training set:")
unique, counts = np.unique(y_train, return_counts=True)
for u, c in zip(unique, counts):
    print(f"  {class_names[u]}: {c} ({c/len(y_train)*100:.1f}%)")
```

```python
plt.figure(figsize=(10, 6))
class_counts = pd.Series(y).value_counts().sort_index()
colors = plt.cm.Set3(np.linspace(0, 1, 7))

bars = plt.bar([class_names[i] for i in class_counts.index],
               class_counts.values, color=colors)

plt.xlabel('Skin Lesion Type', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)
plt.title('Class Distribution in HAM10000 Dataset', fontsize=14)

# Add percentage labels
for bar, count in zip(bars, class_counts.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 50,
             f'{count/len(y)*100:.1f}%', ha='center', fontsize=10)

plt.tight_layout()
plt.savefig('class_distribution.png', dpi=150)
plt.show()

# Imbalance ratio
print(f"\nImbalance ratio (max/min): {class_counts.max()/class_counts.min():.1f}x")
```

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for idx, (class_id, class_name) in enumerate(class_names.items()):
    if idx < 7:
        class_pixels = X_reshaped[y == class_id].flatten()
        axes[idx].hist(class_pixels, bins=50, alpha=0.7, color=colors[idx])
        axes[idx].set_title(f'{class_name} (n={sum(y==class_id)})')
        axes[idx].set_xlabel('Pixel Intensity')
        axes[idx].set_ylabel('Frequency')

axes[7].axis('off')
plt.tight_layout()
plt.savefig('pixel_distribution.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
channel_names = ['Red', 'Green', 'Blue']
channel_colors = ['red', 'green', 'blue']

for i, (name, color) in enumerate(zip(channel_names, channel_colors)):
    for class_id in range(7):
        class_data = X_reshaped[y == class_id][:, :, :, i].flatten()
        axes[i].hist(class_data, bins=50, alpha=0.3, label=class_names[class_id])
    axes[i].set_title(f'{name} Channel Distribution')
    axes[i].set_xlabel('Pixel Intensity')
    axes[i].legend(fontsize=8)

plt.tight_layout()
plt.savefig('color_channels.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 7, figsize=(20, 3))

for idx, (class_id, class_name) in enumerate(class_names.items()):
    mean_image = X_reshaped[y == class_id].mean(axis=0)
    axes[idx].imshow(mean_image)
    axes[idx].set_title(f'{class_name}\n(mean)')
    axes[idx].axis('off')

plt.suptitle('Average Image per Class', fontsize=14, y=1.05)
plt.tight_layout()
plt.savefig('mean_images.png', dpi=150)
plt.show()
```

```python
from sklearn.utils.class_weight import compute_class_weight

# Compute class weights
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = dict(enumerate(class_weights))

print("Class Weights:")
for class_id, weight in class_weight_dict.items():
    print(f"  {class_names[class_id]}: {weight:.3f}")
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Create augmentation generator
train_datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    vertical_flip=True,
    zoom_range=0.1,
    fill_mode='nearest'
)

# For validation - no augmentation
val_datagen = ImageDataGenerator()

# Visualize augmentation
def show_augmented_images(image, generator, n=6):
    fig, axes = plt.subplots(1, n+1, figsize=(15, 3))
    axes[0].imshow(image)
    axes[0].set_title('Original')
    axes[0].axis('off')

    image_batch = np.expand_dims(image, 0)
    aug_iter = generator.flow(image_batch, batch_size=1)

    for i in range(n):
        aug_image = next(aug_iter)[0]
        axes[i+1].imshow(aug_image)
        axes[i+1].set_title(f'Augmented {i+1}')
        axes[i+1].axis('off')

    plt.tight_layout()
    plt.savefig('augmentation_examples.png', dpi=150)
    plt.show()

# Show augmentation on a sample image
sample_idx = np.where(y_train == 4)[0][0]  # melanoma sample
show_augmented_images(X_train[sample_idx], train_datagen)
```

```python
from tensorflow.keras.utils import to_categorical

y_train_encoded = to_categorical(y_train, num_classes=7)
y_test_encoded = to_categorical(y_test, num_classes=7)

print(f"Original label shape: {y_train.shape}")
print(f"Encoded label shape: {y_train_encoded.shape}")
print(f"Sample encoding for class 4 (mel): {y_train_encoded[y_train == 4][0]}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (Conv2D, MaxPooling2D, Dense, Flatten,
                                      Dropout, BatchNormalization, Input)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

def build_cnn_model(input_shape=(28, 28, 3), num_classes=7):
    model = Sequential([
        Input(shape=input_shape),

        # Block 1
        Conv2D(32, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(32, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.25),

        # Block 2
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.30),

        # Block 3
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.40),

        # Classification head
        Flatten(),
        Dense(256, activation='relu'),
        BatchNormalization(),
        Dropout(0.5),
        Dense(128, activation='relu'),
        BatchNormalization(),
        Dropout(0.5),
        Dense(num_classes, activation='softmax')
    ])

    return model

# Build model
model = build_cnn_model()
model.summary()
```

```python
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=15,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=5,
        min_lr=1e-7,
        verbose=1
    ),
    ModelCheckpoint(
        'best_model.keras',
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    )
]
```

```python
# Create data generators
batch_size = 64

train_generator = train_datagen.flow(
    X_train, y_train_encoded,
    batch_size=batch_size
)

# Train the model
history = model.fit(
    train_generator,
    steps_per_epoch=len(X_train) // batch_size,
    epochs=100,
    validation_data=(X_test, y_test_encoded),
    class_weight=class_weight_dict,
    callbacks=callbacks,
    verbose=1
)
```

```python
def plot_training_history(history):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Accuracy
    axes[0].plot(history.history['accuracy'], label='Training', linewidth=2)
    axes[0].plot(history.history['val_accuracy'], label='Validation', linewidth=2)
    axes[0].set_title('Model Accuracy', fontsize=14)
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Accuracy')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    # Loss
    axes[1].plot(history.history['loss'], label='Training', linewidth=2)
    axes[1].plot(history.history['val_loss'], label='Validation', linewidth=2)
    axes[1].set_title('Model Loss', fontsize=14)
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig('training_history.png', dpi=150)
    plt.show()

plot_training_history(history)
```

```python
from sklearn.metrics import (classification_report, confusion_matrix,
                             accuracy_score, precision_recall_fscore_support,
                             roc_auc_score, roc_curve)

# Predictions
y_pred_proba = model.predict(X_test)
y_pred = np.argmax(y_pred_proba, axis=1)

# Overall accuracy
test_accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred,
                           target_names=list(class_names.values())))
```

```python
def plot_confusion_matrix(y_true, y_pred, class_names):
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=list(class_names.values()),
                yticklabels=list(class_names.values()))
    plt.title('Confusion Matrix', fontsize=14)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('True Label', fontsize=12)
    plt.tight_layout()
    plt.savefig('confusion_matrix.png', dpi=150)
    plt.show()

    # Normalized confusion matrix
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    plt.figure(figsize=(10, 8))
    sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=list(class_names.values()),
                yticklabels=list(class_names.values()))
    plt.title('Normalized Confusion Matrix', fontsize=14)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('True Label', fontsize=12)
    plt.tight_layout()
    plt.savefig('confusion_matrix_normalized.png', dpi=150)
    plt.show()

plot_confusion_matrix(y_test, y_pred, class_names)
```
